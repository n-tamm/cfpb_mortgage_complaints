```python
duckdb.sql("""
SELECT 
    company,
    MAX("Date received") AS last_complaint_date
FROM read_parquet('../data/raw/consumer_complaints.parquet')
GROUP BY company
ORDER BY last_complaint_date DESC
""")
```

```python
duckdb.sql("""
SELECT 
    *
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where "Consumer complaint narrative" is not null
order by "Date received" desc
""")
```

```python
duckdb.sql("""
SELECT 
    count(1) as count,
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
           
UNION ALL
           
SELECT count(1)
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
AND "Consumer complaint narrative" IS NOT NULL
""")
```

```python
duckdb.sql("""
SELECT 
    count(1) as count,
    "Sub-product"
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
group by 2 order by 1 desc
""")
```

```python
duckdb.sql("""
SELECT 
    count(1) as count,
    "Sub-product",
    Issue
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
group by 2,3 order by 1 desc
""")
```

```python
duckdb.sql("""
SELECT 
    count(1) as count
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
AND "Consumer disputed?" = 'Yes'
           
UNION ALL
           
SELECT 
    count(1) as count
           
FROM read_parquet('../data/raw/consumer_complaints.parquet')
where Product = 'Mortgage'
AND "Timely response?" = 'No'
""")
```
